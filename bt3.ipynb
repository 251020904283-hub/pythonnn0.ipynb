```python
n = int(input("Enter a positive integer: "))
print(n)

s = 0   # biến dùng để lưu tổng
i = 0   # bắt đầu từ số chẵn đầu tiên

while i < n:
    s += i
    i += 2   # tăng thêm 2 để luôn là số chẵn

print(f"The sum of even numbers less than {n} is {s}")
```
