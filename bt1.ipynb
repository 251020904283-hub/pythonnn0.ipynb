```python
s = input("Enter 3 real number (separated by space):")
s = s.split()
a = float(s[0])
b = float(s[1])
c = float(s[2])
print(f"a = {a} , b = {b}, c = {c}")

# Enter your code from here
import math

# Nhập giá trị của a, b, c
a = float(input("Nhập giá trị a: "))
b = float(input("Nhập giá trị b: "))
c = float(input("Nhập giá trị c: "))

# Kiểm tra nếu a = 0, chuyển thành phương trình bậc nhất
if a == 0:
    if b != 0:
        # Phương trình bậc nhất
        x = -c / b
        print(f"Phương trình bậc nhất có nghiệm: x = {x}")
    else:
        # Nếu a = 0 và b = 0
        if c == 0:
            print("Phương trình có vô số nghiệm.")
        else:
            print("Phương trình vô nghiệm.")
else:
    # Tiến hành giải phương trình bậc hai
    delta = b**2 - 4*a*c
    print(f"Delta = {delta}")

    # Kiểm tra giá trị của delta để phân loại nghiệm
    if delta > 0:
        x1 = (-b + delta**0.5) / (2 * a)
        x2 = (-b - delta**0.5) / (2 * a)
        print(f"Phương trình có hai nghiệm phân biệt: x1 = {x1}, x2 = {x2}")
    elif delta == 0:
        x = -b / (2 * a)
        print(f"Phương trình có nghiệm kép: x = {x}")
    else:
        print("Phương trình vô nghiệm trong tập số thực.")
```
