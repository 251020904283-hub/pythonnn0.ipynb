```python
# Nhập giá trị n từ người dùng
n = int(input("Enter a positive integer: "))

# Khởi tạo biến s để lưu tổng
s = 0

# Dùng vòng lặp for để cộng dồn các số từ 0 đến n
for i in range(n + 1):  # range(n + 1) vì chúng ta muốn tính từ 0 đến n
    s += i  # Cộng dồn vào biến s

# Hiển thị kết quả
print(f"The sum of 1 to {n} is {s}")
```
